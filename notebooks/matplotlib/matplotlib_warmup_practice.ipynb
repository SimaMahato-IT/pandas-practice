1. Linear/Line Chart

```python
import matplotlib.pyplot as plt
```

```python
months=[1,2,3,4]
sales=[1000,1500,1200,1800]
plt.plot(months,sales)
plt.show()
```

```python
months=[1,2,3,4]
sales=[1000,1500,1200,1800]
plt.plot(months,sales,color='blue', linestyle='--',linewidth='2', marker='o', label='2026 sales data')
plt.xlabel('Months')
plt.ylabel('Sales per month')
plt.title('Monthly Sales Data Report')

plt.show()
```

```python
months=[1,2,3,4]
sales=[1000,1500,1200,1800]
plt.plot(months,sales,color='blue', linestyle='--',linewidth='2', marker='o', label='2026 sales data')
plt.xlabel('Months')
plt.ylabel('Sales per month')
plt.title('Monthly Sales Data Report')
plt.legend(loc='upper left', fontsize=12)
plt.grid(color='gray',linewidth=1)
plt.show()
```

```python
months=[1,2,3,4]
sales=[1000,1500,1200,1800]
plt.plot(months,sales,color='green', linestyle='--',linewidth='2', marker='o', label='2026 sales data')
plt.xlabel('Months')
plt.ylabel('Sales per month')
plt.title('Monthly Sales Data Report')
plt.legend(loc='upper left', fontsize=12)
plt.grid(color='gray',linestyle=':', linewidth=1)
plt.show()
```

```python
months=[1,2,3,4]
sales=[1000,1500,1200,1800]
plt.plot(months,sales,color='green', linestyle='--',linewidth='2', marker='o', label='2026 sales data')
plt.xlabel('Months')
plt.ylabel('Sales per month')
plt.title('Monthly Sales Data Report')
plt.legend(loc='upper left', fontsize=12)
plt.grid(color='gray',linestyle=':', linewidth=1)
plt.xlim(1,4)
plt.ylim(0,2000)
plt.show()
```

2. Bar Chart

```python
product=['A','B','C','D']
sales=[1000,1500,800,1200]
plt.bar(product,sales)
plt.show()
```

```python
product=['A','B','C','D']
sales=[1000,1500,800,1200]
plt.bar(product,sales, color='green', label='Sales 2026')
plt.xlabel('Product')
plt.ylabel('Sales')
plt.title('Product Sales Comparison')
plt.legend()
plt.show()
```

# Horizontal Bar Chart

```python
product=['A','B','C','D']
sales=[1000,1500,800,1200]
plt.barh(product,sales, color='green', label='Sales 2026')
plt.xlabel('Product')
plt.ylabel('Sales')
plt.title('Product Sales Comparison')
plt.legend()
plt.show()
```

# 2.Pie Chart(Slice of a Circle)

```python
regions=['North','South','East','West']
revenue=[3000,2000,1500,1000]
plt.pie(revenue,labels=regions, autopct='%1.1f%%',)
plt.title('Revenue Cpntribution By Region')
plt.show()
```

```python
regions=['North','South','East','West']
revenue=[3000,2000,1500,1000]
c=['gold','skyblue','lightgreen','coral']
plt.pie(revenue,labels=regions, autopct='%1.1f%%', colors=c)
plt.title('Revenue Cpntribution By Region')
plt.show()
```

# 4. Histogram Chart

```python
scores=[45,67,89,78,88,92,60,74,81,59,66,75,82,90,85,70,73,68,77]
plt.hist(scores, bins=5, color= 'skyblue', edgecolor='black')


plt.show()
```

```python
scores=[45,67,89,78,88,92,60,74,81,59,66,75,82,90,85,70,73,68,77]
plt.hist(scores, bins=5, color= 'skyblue', edgecolor='black',label='Class IX Score Graph')
plt.xlabel('Score Range')
plt.ylabel('Number of students')
plt.title('Score Distribution of Students')
plt.legend()
plt.show()
```

# 5. Scatter Plot

```python
hours_studies =[1,2,3,4,5,6,7,8]
exam_scores=[50,55,60,70,75,80,85,90]
plt.scatter(hours_studies, exam_scores, color='green', marker='o',label='Student Data')
plt.xlabel('Hours Studied')
plt.ylabel('Exam Score')
plt.title('Relationship between Study Time & Exam Score')
plt.legend()
plt.grid(True)
plt.show()
```

# Subplots

```python
x=[1,2,3,4]
y=[10,20,15,25]
plt.subplot(1,2,1) #1at row,2nd col, 1st subplot
plt.plot(x,y)
plt.title('Line Chart')
plt.subplot(1,2,2) #1at row,2nd col, 2nd subplot
plt.bar(x,y)
plt.title('Bar Chart')
plt.tight_layout()
plt.show()
```

```python
fig, ax=plt.subplots(1,2, figsize=(10,5))

x=[1,2,3,4]
y=[10,20,15,25]
ax[0].plot(x,y) 
ax[0].set_title('Line Chart')
ax[1].bar(x,y)
ax[1].set_title('Bar Chart')


plt.tight_layout()
plt.show()
```
